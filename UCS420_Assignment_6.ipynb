Question 1

```python
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 200)

y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y1, color='blue', linestyle='-', label='y = M*x²')
plt.plot(x, y2, color='red', linestyle='--', label='y = M*sin(x)')

plt.title("Mathematical Functions")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()
plt.grid(True)
plt.show()
```

Question 2

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Subject': ['Python', 'Java', 'Maths', 'C++', 'DBMS'],
    'Score': [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)

print(df)

plt.figure(figsize=(8, 5))
ax = sns.barplot(data=df, x='Subject', y='Score', hue='Subject', palette='Set2', legend=False)

for container in ax.containers:
    ax.bar_label(container)

plt.title("Scores in Five Subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(12345)

data = np.random.randn(50)
x = np.arange(1, 51)
noise = np.random.randn(50)

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(x, data.cumsum(), color='blue')
ax[0, 0].set_title("Cumulative Sum")
ax[0, 0].set_xlabel("Index")
ax[0, 0].set_ylabel("Cumulative Value")
ax[0, 0].grid(True)

ax[0, 1].scatter(data, noise, color='red')
ax[0, 1].set_title("Scatter Plot with Random Noise")
ax[0, 1].set_xlabel("Random Values")
ax[0, 1].set_ylabel("Noise")
ax[0, 1].grid(True)

ax[1, 0].plot(x, data, color='green', linestyle='--')
ax[1, 0].set_title("Random Values Line Plot")
ax[1, 0].set_xlabel("Index")
ax[1, 0].set_ylabel("Value")
ax[1, 0].grid(True)

ax[1, 1].hist(data, bins=10, color='orange', edgecolor='black')
ax[1, 1].set_title("Distribution of Random Values")
ax[1, 1].set_xlabel("Value")
ax[1, 1].set_ylabel("Frequency")
ax[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)

print(df.head())

sns.set_style("whitegrid")

plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='month_number', y='total_profit', marker='o')
plt.title("Total Profit of All Months")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()

product_columns = [
    'facecream', 'facewash', 'toothpaste',
    'bathingsoap', 'shampoo', 'moisturizer'
]

plt.figure(figsize=(12, 6))

for column in product_columns:
    sns.lineplot(data=df, x='month_number', y=column, marker='o', label=column)

plt.title("Monthly Sales of All Products")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()

numeric_columns = df.select_dtypes(include='number').columns

for column in numeric_columns:
    plt.figure(figsize=(10, 5))
    sns.barplot(data=df, x='month_number', y=column, color='skyblue')
    plt.title(f"Bar Chart of {column}")
    plt.xlabel("Month")
    plt.ylabel(column)
    plt.grid(axis='y')
    plt.show()
```
