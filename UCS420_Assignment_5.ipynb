```python
import numpy as np

x = np.array([1,2,3,4,5,1,2,1,1,1])
values, counts = np.unique(x, return_counts = True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most Frequent Value: ", most_frequent)
print("Frequency: ", np.max(counts))
print("Indices: ", indices)


y = np.array([1,1,1,2,3,4,2,4,3,3])
values,counts = np.unique(y, return_counts = True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most Frequent Value: ", most_frequent)
print("Frequency: ", np.max(counts))
print("Indices: ", indices)
```

Question 4

```python
import numpy as np

lavdeep = np.linspace(10, 100, 25)

print("Array:", lavdeep)
print("Dimensions:", lavdeep.ndim)
print("Shape:", lavdeep.shape)
print("Total Elements:", lavdeep.size)
print("Data Type:", lavdeep.dtype)
print("Total Bytes:", lavdeep.nbytes)

transposed = lavdeep.reshape(25, 1)
print("Transpose using reshape():\n", transposed)

print("Transpose using T attribute:\n", lavdeep.T)
```

```python
import numpy as np

ucs420_lavdeep = np.array([[10, 20, 30, 40],
                           [50, 60, 70, 80],
                           [90, 15, 20, 35]])

print("Original Array:\n", ucs420_lavdeep)
print("Mean:", np.mean(ucs420_lavdeep))
print("Median:", np.median(ucs420_lavdeep))
print("Maximum:", np.max(ucs420_lavdeep))
print("Minimum:", np.min(ucs420_lavdeep))
print("Unique Elements:", np.unique(ucs420_lavdeep))

reshaped_ucs420_lavdeep = ucs420_lavdeep.reshape(4, 3)
print("Reshaped Array (4x3):\n", reshaped_ucs420_lavdeep)

resized_ucs420_lavdeep = np.resize(ucs420_lavdeep, (2, 3))
print("Resized Array (2x3):\n", resized_ucs420_lavdeep)
```
