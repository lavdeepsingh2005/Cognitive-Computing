Assignment 5.3 [Prime]

```python
n = int(input("Enter a Number: "))

s = 0

for i in range (2, n + 1) :
  prime = True

  for j in range (2, i) :
    if i % j == 0 :
      prime = False
      break

  if prime :
    s = s + i

print("Sum of Prime Numbers:",s)
```

Assignment 6.1 [Function]

```python
def AddOddNumbers(n) :
  s = 0

  for i in range(1, n + 1) :
    if i % 2 != 0 :
      s = s + i
  return s

n = int(input("Enter a Number: "))
print("Sum of Odd Numbers: ", AddOddNumbers(n))
```

Assignment 6.2 [Function]

```python
def AddPrimeNumbers(n) :
  s = 0

  for i in range (2, n + 1) :
    prime = True

    for j in range (2, i) :
      if i % j == 0 :
        prime = False
        break

    if prime :
      s = s + i

  return s

n = int(input("Enter a Number: "))
print("Sum of Prime Numbers: ", AddPrimeNumbers(n))
```
