Question 1 [List]

```python
from ast import ListComp
L = [1,0,2,4,1,7,0,1,9,7]
print("Original List: ", L)

newL = [i * 10 for i in L]
print("After Multiplication: ", newL)

L.append(1)
L.append(8)
print("After Adding: ",L)
#After using append, the number inside the brackets was added at the end of the list

L.insert(3,7)
print("After Inserting: ", L)
#After using insert, the number entered after the comma was added at the position mentioned before the comma inside the brackets

L.remove(7)
print("After Removal: ", L)

L.pop()
print("After Popping: ", L)

L.sort()
print("Sorting in Ascending Order: ", L)

L.sort(reverse=True)
print("Sorting in Descending Order: ", L)

print("First 3 Elements: ", L[:3])
print("Last 3 Elements: ", L[-3:])

L_avg = sum(L) / len(L)
print("Average: ", L_avg)

newL = [i for i in L if i > L_avg]
print("Elements Greater Than Average: ", newL)
```

Question 3 [Random Numbers]

```python
import random as r

r.seed(1024170197)

numbers = [r.randint(100,900) for _ in range(100)]
print("100 Random Numbers between 100 and 900: ", numbers)

even_numbers = [i for i in numbers if i % 2 == 0]
print("Count (Even Numbers): ", len(even_numbers))
print("Even Numbers: ", even_numbers)

odd_numbers = [i for i in numbers if i % 2 != 0]
print("Count (Odd Numbers): ", len(odd_numbers))
print("Odd Numbers: ", odd_numbers)

def Prime(n) :
  if n < 2 :
    return False
  for i in range(2, int(n**0.5) + 1) :
    if n % i == 0 :
      return False
  return True

prime_numbers = [i for i in numbers if Prime(i)]
print("Count (Prime Numbers): ", len(prime_numbers))
print("Prime Numbers: ", prime_numbers)

frequency = {}

for i in numbers :
  if i in frequency :
    frequency[i] += 1
  else :
    frequency[i] = 1

most_frequent_number = max(frequency, key=frequency.get)
print("Most Frequent Number: ", most_frequent_number)
print("Number of Occurences: ", frequency[most_frequent_number])
```

Question 4 [Set]

```python
roll_number = [1,0,2,4,1,7,0,1]

A = {roll_number * 7 for roll_number in roll_number}
B = {roll_number * 9 for roll_number in roll_number}

print("Set A: ", A)
print("Set B: ", B)

union = A.union(B)
print("Union of A and B: ", union)

intersection = A.intersection(B)
print("Intersection of A and B: ", intersection)

difference = A.difference(B)
print("A - B: ", difference)

difference2 = B.difference(A)
print("B - A: ", difference2)

symmetric_difference = A.symmetric_difference(B)
print("Symmetric Difference of A and B: ", symmetric_difference)

#Difference provides the elements present in one set and not in the other
#Symmetric difference provides us with elements that are present in either set but not in both

print("Is A a subset of B? ", A.issubset(B))
print("Is B a subset of A? ", B.issubset(A))

X = int(input("Enter a value: "))
A.discard(X)
print("Set A after discard: ", A)
```

Question 5 [Dictionary]

```python
my_dict = {
    "name": "Lavdeep Singh",
    "roll_no": "1024170197",
    "branch": "COPC",
    "age": 20,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)

my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city:")
print(my_dict)

my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)

my_dict["age"] += 1

print("\nAfter increasing age:")
print(my_dict)

dict_pop = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_branch)

dict_del = my_dict.copy()

del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

print("\nKey → Value:")

for key, value in my_dict.items():
    print(key, "→", value)


if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")


friend_dict = {
    "name": "Kamlesh",
    "roll_no": "1023833",
    "branch": "CSE",
    "age": 19,
    "city": "Mumbai"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nString values only:")
print(string_values)
```
